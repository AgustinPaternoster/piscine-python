```python
import matplotlib.pyplot as plt

# Creamos la figura y el gráfico
fig, ax = plt.subplots()

# Datos
meses = ['Ene', 'Feb', 'Mar', 'Abr']
ventas = [10, 15, 7, 19]

# Dibujamos la línea de color verde, discontinua y con marcadores circulares
ax.plot(meses, ventas, color='green', linestyle='--', marker='o')

# Añadimos las etiquetas y el título
ax.set_title('Evolución de Ventas del Primer Trimestre')
ax.set_xlabel('Meses')
ax.set_ylabel('Cantidad de Ventas')

# Mostramos el resultado
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

meses = ['Ene', 'Feb', 'Mar', 'Abr']
ventas_A = [10, 15, 7, 19]
ventas_B = [5, 11, 14, 12]

# Primera línea: añadimos el label 'Producto A'
ax.plot(meses, ventas_A, color='blue', marker='o', label='Producto A')

# Segunda línea: añadimos el label 'Producto B' sobre el MISMO objeto 'ax'
ax.plot(meses, ventas_B, color='orange', marker='s', label='Producto B')

# Configuramos títulos y textos
ax.set_title('Comparativa de Ventas')
ax.set_xlabel('Meses')
ax.set_ylabel('Unidades Vendidas')

# ¡CRUCIAL! Si no llamas a este método, las etiquetas ('label') no se mostrarán
ax.legend()

plt.show()
```
