```python
import pandas as pd
import matplotlib.pyplot as plt
from load_csv import load

country_1 = 'Spain'
country_2 = 'France'
path = "population_total.csv"

def format(valor):
    multipliers = {'m': 1000000 , 'k': 1000, 'b':1000000000}
    if valor.isdecimal() or pd.isna(valor):
        return valor
    return int((float(valor[:-1]) * multipliers[valor[-1].lower()]))

def render_graph(country_1: str , country_2: str, file_path:str):
    df = load("population_total.csv")
    df_filtered = df.set_index('country').loc[[country_1, country_2], :'2050']
    country_1_serie = df_filtered.iloc[0]
    country_2_serie = df_filtered.iloc[1]
    return country_1_serie, country_2_serie
```
