```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from load_csv import load


gdp = load('./income_per_person_gdppercapita_ppp_inflation_adjusted.csv')
life= load('../life_expectancy_years.csv')
df_merge = pd.merge(gdp[['country', '1900']], gdp[['country', '1900']], on="country", suffixes=("_gdp", "_years"))
df_clean = df_merge.dropna().rename(columns={'1900_x':'gdp_per_capita','1900_y':'years'})
df_clean




```

```python

def plot_format(x, pos):
    if x >= 1000000000:
        return f'{x/1000000000:.1f}B'
    elif x >= 1000000:
        return f'{x/1000000:.0f}M'
    elif x >= 1000:
        return f'{x/1000:.0f}k'
    else:
        return f'{x:.0f}'


fig , ax = plt.subplots()

ax.set_xlim(290,11000)
ax.set_xscale('log')
valores = [300, 1000, 10000]
ax.set_xticks(valores)
ax.xaxis.set_major_formatter(ticker.FuncFormatter(plot_format))

ax.scatter(df_clean['gdp_per_capita'], df_clean['years'], s=40)
```
